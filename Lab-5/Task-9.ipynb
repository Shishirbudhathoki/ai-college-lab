#### Save the trained model's state_dict, reload it into a new model instance, and verify predictions match the original model.

```python
import torch
from torch import nn
```

```python
from torch import nn

class NeuralNetwork(nn.Module):
    def __init__(self):
        super().__init__()
        self.layer_1 = nn.Linear(in_features=2,out_features=10)
        self.layer_2 = nn.Linear(in_features=10,out_features=10)
        self.layer_3 = nn.Linear(in_features=10, out_features=1)
        self.relu = nn.ReLU() 

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        return self.layer_3(self.relu(self.layer_2(self.relu(self.layer_1(x)))))
        
```

```python
torch.manual_seed(42)
model = NeuralNetwork()
model.state_dict()
```

```python
model.load_state_dict(
    torch.load("model.pth")
)
```

```python
model.eval()
```

```python
x = torch.tensor([
    [0.5, 0.5]
], dtype=torch.float)


with torch.inference_mode():
    logit = model(x)
    probability = torch.sigmoid(logit)
    prediction = torch.round(probability)


print("Logit:", logit.item())
print("Probability:", probability.item())
print("Prediction:", prediction.item())
```
