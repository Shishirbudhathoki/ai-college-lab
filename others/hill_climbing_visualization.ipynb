
### Simple Hill Climbing Search with Visualization


```python
import random
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Objective function
def objective_function(x):
    return -x**2 + 5*x
```

```python
# Hill Climbing Algorithm
def hill_climbing_with_visualization(objective_function, x_start, step_size, max_iterations):
    current_x = x_start
    current_value = objective_function(current_x)

    # Store path for visualization
    x_path = [current_x]
    value_path = [current_value]

    for i in range(max_iterations):
        # Generate a neighbor
        next_x = current_x + random.choice([-step_size, step_size])
        next_value = objective_function(next_x)

        # If the neighbor is better, move to it
        if next_value > current_value:
            current_x = next_x
            current_value = next_value
        
        x_path.append(current_x)
        value_path.append(current_value)
    
    return x_path, value_path, current_x, current_value
```

```python
# Visualization Function
def visualize_hill_climbing(x_path, value_path, objective_function):
    x = np.linspace(-5, 5, 1000)
    y = [objective_function(i) for i in x]

    plt.figure(figsize=(10, 6))
    plt.plot(x, y, label='Objective Function', color='blue')
    plt.scatter(x_path, value_path, color='red', label='Hill Climbing Path')
    plt.plot(x_path, value_path, linestyle='--', color='red')
    plt.title('Hill Climbing Search Visualization')
    plt.grid()
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.legend()
    plt.show()
```

```python
x_start = random.uniform(-5, 5)  # Random starting point
step_size = 0.1
max_iterations = 100

# Run Hill Climbing Algorithm
x_path, value_path, best_x, best_value = hill_climbing_with_visualization(objective_function, x_start, step_size, max_iterations)

# Visualize the Results
visualize_hill_climbing(x_path, value_path, objective_function)
print(f"Started at: x = {x_start}, f(x) = {objective_function(x_start)}")

print(f"Best solution found: x = {best_x}, f(x) = {best_value}")
```
